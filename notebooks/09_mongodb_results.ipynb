# EarthScape Climate Agency — Notebook 09: MongoDB Storage & Results

### Objective:
Verify the full database schema in MongoDB, query the indexed collections (`users`, `weather_events_cleaned`, `climate_summary`, `anomalies`, `predictions`, `alerts`, `system_logs`), and display application-ready payloads.


## 1. Connect to MongoDB Instance


```python
from pymongo import MongoClient
import pandas as pd

client = MongoClient('mongodb://localhost:27017/')
db = client['earthscape_climate_db']
print("Connected to MongoDB:", db.name)
print("Available Collections:", db.list_collection_names())
```

## 2. Inspect KPI Summary Payload


```python
summary_doc = db['climate_summary'].find_one()
if summary_doc:
    print("KPIs:", summary_doc.get('kpis'))
else:
    print("No summary doc found yet.")
```

## 3. Inspect Anomalies Collection


```python
anom_sample = list(db['anomalies'].find({}, {'_id': 0, 'EventId': 1, 'Type': 1, 'Severity': 1, 'State': 1, 'anomaly_score': 1}).limit(5))
pd.DataFrame(anom_sample)
```

## 4. Inspect Forecast Predictions Payload


```python
pred_doc = db['predictions'].find_one()
if pred_doc:
    print("Metrics:", pred_doc.get('metrics'))
    display(pd.DataFrame(pred_doc.get('series', [])).head(5))
```

## 5. Inspect Active Climate Alerts


```python
alerts = list(db['alerts'].find({}, {'_id': 0}).limit(5))
pd.DataFrame(alerts)
```

### Conclusion:
MongoDB contains all clean records, ML anomaly outputs, forecast series, and triggered alert collections fully indexed and ready for sub-second Streamlit dashboard retrieval.
